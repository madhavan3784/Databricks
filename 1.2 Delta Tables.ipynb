```python
%sql
create table employee (id int, name string, salary double)
```

```python
%sql
insert into employee 
values 
(1, 'adam', 23000),
(3, 'jen', 25000),
(2, 'john', 40000),
(4, 'mike', 50000),
(5, 'jerry', 35000)
```

```python
%sql
select * from employee
```

```python
%sql
update employee set salary = salary + 1000 where id >=4;
    
select * from employee
```
